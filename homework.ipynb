**Concept:** *reading CSV data, matplotlib, line graph*

**Task:**

Write a function `read_and_plot_column(filename, column_name)` that:
* Opens a CSV file (do **not** use pandas)
* Extracts the values from the specified column
* Casts all values to `float`
* Plots the values using `plt.plot()`
* Adds a title
* Displays the plot

````python
def read_and_plot_column(filename, column_name):
    """Read and plot a CSV column."""

    # Find column position
    with open(filename) as file:
        columns = file.readline().strip().split(",")

    # Read numeric values
    values = np.genfromtxt(
        filename,
        delimiter=",",
        skip_header=1,
        usecols=columns.index(column_name)
    )

    plt.plot(values)
    plt.title(column_name)
    plt.show()
````

**Task:**

Using `df_weather`:
* Write a helper function to extract time (in hours) from a datetime string and convert it to decimal hours (float)
* Convert time using:

    `hours + (minutes / 60) + (seconds / 3600)`
* Create a new DataFrame with the same index as `df_weather`
* Add the following columns:
  * `sunrise` (converted to float)
  * `sunset` (converted to float)
  * `tempmin` (copied directly from `df_weather`)
* Plot all columns together using `.plot()`


**Sample Output:**

This is similar to how your new dataframe should look before plotting:

```
   start_time  end_time  price
P1       8.25     17.50     20
P2       8.50     17.75     35
P3       9.00     18.00     15
P4       9.25     18.25     60
P5       9.50     18.50     50
```

````python
def time_to_hours(datetime_str):
    """Convert datetime to decimal hours."""
    time = datetime_str.split("T")[1]
    hours, minutes, seconds = map(int, time.split(":"))
    return hours + minutes / 60 + seconds / 3600

# Create new DataFrame
df_sun = pd.DataFrame(index=df_weather.index)

# Add columns
df_sun["sunrise"] = df_weather["sunrise"].apply(time_to_hours)
df_sun["sunset"] = df_weather["sunset"].apply(time_to_hours)
df_sun["tempmin"] = df_weather["tempmin"]

# Plot all columns
df_sun.plot()
plt.title("Sunrise, Sunset, and Minimum Temperature")
plt.show()
````

**Task:**


Write a function `filter_rows_matching_pattern(df, column, pattern)` that:
* Returns rows where the specified column contains the given pattern
* Handles missing values
* Prints the number of matching rows

**Sample Output:**

If your selected column contained matching text values, the following could be your output:

````python
# Your Solution
def filter_rows_matching_pattern(df, column, pattern):
    """Filter rows containing a given pattern."""

    # Find matching rows, ignoring missing values
    filtered_df = df[df[column].str.contains(pattern, na=False)]

    # Print number of matches
    print(f"Found {len(filtered_df)} records of {column} containing {pattern}.")

    return filtered_df
````

Write a function `create_df_categorical(df_source, column_name)` that:


````python
# Your Solution
def create_df_categorical(df_source, column_name):
    """Create a DataFrame with category counts and percentages."""

    # Count each category
    counts = df_source[column_name].value_counts()

    # Calculate percentages
    percentages = df_source[column_name].value_counts(normalize=True)

    # Create new DataFrame
    df_categorical = pd.DataFrame({
        "count": counts,
        "pct": percentages
    })

    return df_categorical
````

**Problem 17**

**Concept:** *Filtering records by category.*


**Task:**

Write a function `filter_weather(df, conditions, temp_min, temp_max)` that:
* Filters rows where:
  * conditions contain the given string
  * `tempmax` ≤ `temp_max`
  * `tempmin` ≥ `temp_min`
* Prints intermediate counts
* Returns the filtered DataFrame

**Sample Output:**

If your data matched the filtering conditions step by step, the following could be your output:
```
4 records with conditions Blue
3 records with conditions Blue and tempmax <= 50
2 records with conditions Blue and 20 <= temp <= 50
Found 2 records with conditions of Blue and temp between 20 and 50.
```

````python
# Your Solution
def filter_weather(df, conditions, temp_min, temp_max):
    """Filter weather by conditions and temperature."""

    # Filter by conditions
    filtered = df[df["conditions"].str.contains(conditions, na=False)]
    print(f"{len(filtered)} records with conditions {conditions}")

    # Filter by maximum temperature
    filtered = filtered[filtered["tempmax"] <= temp_max]
    print(f"{len(filtered)} records with conditions {conditions} and tempmax <= {temp_max}")

    # Filter by minimum temperature
    filtered = filtered[filtered["tempmin"] >= temp_min]
    print(f"{len(filtered)} records with conditions {conditions} and {temp_min} <= temp <= {temp_max}")

    print(f"Found {len(filtered)} records with conditions of {conditions} and temp between {temp_min} and {temp_max}.")

    return filtered
````

**Problem 18**

**Concept:** *Creating DataFrame, standardizing values.*


**Task:**

Write a function `create_standardized_values(df_source, columns, row_index=None)` that:
* Creates a new DataFrame
* Standardizes values relative to a given row
* Uses the first row if none is provided
* Returns the new DataFrame

**Sample Output:**

If the following was your dataframe:
```
      high_value  low_value
2020        100        50
2021        120        55
2022        140        60
2023        160        70
2024        180        75
```
If you call the function
`create_standardized_values(df_example, ["high_value", "low_value"], 2020)`,
you should get the following as the output:
```
      high_value  low_value
2020         1.0       1.00
2021         1.2       1.10
2022         1.4       1.20
2023         1.6       1.40
2024         1.8       1.50

```

````python
# Your Solution
def create_standardized_values(df_source, columns, row_index=None):
    """Standardize columns relative to a selected row."""

    # Select columns
    df_new = df_source[columns].copy()

    # Choose reference row
    if row_index is None:
        reference = df_new.iloc[0]
    else:
        reference = df_new.loc[row_index]

    # Divide by reference values
    df_new = df_new / reference

    return df_new
````
